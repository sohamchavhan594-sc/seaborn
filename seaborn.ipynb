```python
# import the required libraries
import matplotlib.pyplot as plt
import seaborn as sns 
import pandas as pd
import numpy as np
```

```python
# getting the built - in datasets of seaborn.
sns.get_dataset_names()
```

```python
sns.load_dataset('iris')
```

```python
df=sns.load_dataset('iris')
```

```python
df.species.unique()
```

```python
df.head()
```

```python
# Line Plot()
```

```python

sns.lineplot(x='sepal_length',y='sepal_width',data=df, ci=None)                    #Confidence interval
plt.show()
```

```python
sns.lineplot(x='sepal_length',y='sepal_width',data=df,hue='species', ci=None)
plt.show()
```

```python
# Scatter plot()
```

```python
sns.scatterplot(x='sepal_length',y='sepal_width', data=df,hue='species')
plt.title("scatter plot")
plt.show()
```

```python
# Bar plot()
sns.barplot(x='species',y='sepal_length',data=df)
plt.show()
```

```python
# Count plot()
sns.countplot(x='species',data=df)
plt.show()                                                          # Acount plot basically counts the categories and returns a
                                                                    # count of their occurrence.
```

```python
# Boxplot()
sns.boxplot(x='sepal_width',data=df)                                    # Presewnce of outlayers
plt.show
```

```python
sns. boxplot(x='species', y='sepal_width',data=df)
plt.show()
# Checking outliers in each category with respect to sepal width
```

```python
# Violinplot()
# Dencity estimate to give a better description about the data distribution.
sns.violinplot(x='species',y='sepal_width',data=df)
plt.show()
```

```python
# Stripplot()
#It basically creats a scatter plot based on the category.
# It is created using the stripplot()
sns.stripplot(x='species',y='petal_width', data=df)
plt.show()
```

```python
# Histplot()
#It is accuratew method for the graphical representation of numeric data
# disatribution
sns.histplot(x='petal_width', data=df)
plt.show()
```

```python
# kde is appromexition of histogram to densityu curve.
sns.histplot(x='petal_width', data=df,hue='species', kde=True)
plt.show()
```

```python
# Distplot()
# Distpot is used baasically for unvariant set of observations
#and visualize it through a histogram
# i.e. only one observation and hence we choose one partiocular column of the dataset.
sns.distplot(df['petal_width'], color='g')    # hist=False , kde = False
plt.show()
# Density means ther probability of given point to lie on the curve.
```

```python
# Pairplot()
sns.pairplot(data= df, hue= 'species')
plt.show()
```

```python
# categorical plot
sns.catplot(x= 'sepal_length', data=df, kind='violin')
plt.show()
```

```python
# The default representation of the data in catplot () use a stripplot
sns.catplot(x='species', y='sepal_length', data=df, kind='swarm')
plt.show()
```

```python
sns.catplot(x='petal_length', y= 'species',data=df)
plt.show()
```

```python
# Heat map
# corretion between the different parameters
tc = df.corr(numeric_only=True)
tc
```
